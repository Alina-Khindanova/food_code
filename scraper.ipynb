```python
## Local Functions

def safely_close_popup(id_list:list):
    """
    Closes pop up with id name in id_list. You can iterate through different pop up id names.
    Parameters
    ===========
    :param id_list: a list with strings
    """
    
    for id_name in id_list:
        try:
            pointer = driver.find_element(By.ID, id_name)
            pointer.click()
            print(f'Popup with id name {id_name} was closed!')
        except:
            print(f'Id {id_name} was not found!')
            
```

```python
wait = WebDriverWait(driver, 10)

for t in names[:10]:
    d = t.text
    print(d)
    try:
        safely_close_popup(['poptinFormCloseText'])
        element = wait.until(EC.presence_of_element_located((By.LINK_TEXT, d)))

        # Scroll the element into view
        driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", element)
        time.sleep(0.5)

        # Try clicking
        element.click()
        print("Clicked successfully:", d)

    except Exception as e:
        print("Click failed for:", d, "| Retrying with JS click")
        # Fallback to JavaScript click
        driver.execute_script("arguments[0].click();", element)
        print("Clicked with JS:", d)
    driver.back()
    time.sleep(2)

# driver.close()
```
