```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.common.exceptions import NoSuchElementException
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import csv
import time
```

```python
options = Options()
#options.add_argument("--headless")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
```

```python
BASE_URL = "https://www.ewg.org/foodscores/products/?category_group=Baby+Food"
driver.get(BASE_URL)
time.sleep(3)
```

```python
try:
    # Adjust the selector to your popup's close button
    popup_close = driver.find_element(By.XPATH, '//*[@id="poptinFormCloseText"]/div/p/span/span/u/strong')  # Example class
    popup_close.click()
    print("Popup closed")
except NoSuchElementException:
    print("No popup found")
except:
    print("Popup could not be closed")

```

```python
wait = WebDriverWait(driver, 10)

for t in names:
    d = t.text
    print(d)
    try:
        element = wait.until(EC.presence_of_element_located((By.LINK_TEXT, d)))

        # Scroll the element into view
        driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", element)
        time.sleep(0.5)

        # Try clicking
        element.click()
        print("Clicked successfully:", d)

    except Exception as e:
        print("Click failed for:", d, "| Retrying with JS click")
        # Fallback to JavaScript click
        driver.execute_script("arguments[0].click();", element)
        print("Clicked with JS:", d)
    driver.back()
    time.sleep(2)

driver.close()
```
